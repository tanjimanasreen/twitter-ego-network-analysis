# Utility Functions


```python
def load_json(df):
    
    result = df.to_json(orient="index")
    parsed = json.loads(result)
    data = json.dumps(parsed, indent=4)
    return data


def create_dummy_db():
    
    path = "English/Unzipped/Schizophrenia_eng/"

    file_list = os.listdir(path)

    df_append = pd.DataFrame()
    #append all files together
    for file in file_list:
        df_temp = pd.read_csv(path + file)
        df_append = df_append.append(df_temp, ignore_index=True)

    return df_append
```

# Phase 1 
### Construction of Dataset

```python
def get_ids(topic="Schizophrenia"): # Specify the specific group of users
    
    df = pd.read_csv("English/user_features.csv")
    
    data = load_json(df) # Connect and Load Database
    data = ast.literal_eval(data)
    
    id_list = []
    for key, value in data.items():
        if data[key]['diagnosis'] == topic:
            id_list.append(data[key]['user_id'])
    user_ids = json.dumps(id_list)
     
    return user_ids


def get_twitter_usernames(topic="Schizophrenia"):
    
    df = pd.read_csv("English/user_features.csv") 
    
    data = load_json(df) # Connect and Load Database
    data = ast.literal_eval(data)
    
    username_list = []
    for key, value in data.items():
        if data[key]['diagnosis'] == topic:
            username_list.append(data[key]['user_screen_name'])
    twitter_usernames = json.dumps(username_list)
    
    return twitter_usernames


def getTweetsByID():
    
    # Connect and Load Data
    data = load_json(create_dummy_db()) 
    df_data = pd.read_json(data, orient ='index')
    
    # Get the specific user_ids
    user_ids = ast.literal_eval(get_ids())
    target_users = user_ids[:10]
    
    # Get all the tweets of specific users
    tweets_df = df_data.loc[df_data['user_id'].isin(target_users)]
    
    return tweets_df

getTweetsByID()
```
